1. Qual é a prevalência de risco/diagnóstico de diabetes dividida por faixas etárias (jovens, adultos, idosos)?


```python
%sql
SELECT 
    faixa_etaria,
    COUNT(*) AS total_pacientes,
    SUM(diabetes) AS total_casos,
    ROUND(AVG(diabetes) * 100, 2) AS pct_prevalencia
FROM mvp.staging.gold_fato_diabetes
GROUP BY faixa_etaria
ORDER BY pct_prevalencia DESC;
```

2. Como o Índice de Massa Corporal (IMC) e os níveis de glicose em jejum correlacionam-se com a presença da doença?

```python
%sql
SELECT 
    categoria_bmi,
    categoria_glicose,
    COUNT(*) AS total_pacientes,
    SUM(diabetes) AS casos_diabetes,
    ROUND(AVG(diabetes) * 100, 2) AS pct_prevalencia
FROM mvp.staging.gold_fato_diabetes
GROUP BY categoria_bmi, categoria_glicose
ORDER BY categoria_bmi, categoria_glicose;
```

3. Pacientes com histórico familiar de diabetes ou com sedentarismo possuem uma taxa proporcionalmente maior de risco?

```python
%sql
SELECT 
    family_history_diabetes,
    physical_activity_level,
    COUNT(*) AS total_pacientes,
    SUM(diabetes) AS casos_diabetes,
    ROUND(AVG(diabetes) * 100, 2) AS pct_prevalencia
FROM mvp.staging.gold_fato_diabetes
GROUP BY family_history_diabetes, physical_activity_level
ORDER BY family_history_diabetes DESC, pct_prevalencia DESC;
```

4. Qual é a média de nível de glicose e IMC entre pacientes diagnosticados vs. não diagnosticados?

```python
%sql
SELECT 
    CASE WHEN diabetes = 1 THEN 'Diagnosticado (Diabetes)' ELSE 'Não Diagnosticado' END AS status_paciente,
    COUNT(*) AS total_pacientes,
    ROUND(AVG(fasting_blood_sugar), 2) AS media_glicose_jejum,
    ROUND(AVG(bmi), 2) AS media_imc
FROM mvp.staging.gold_fato_diabetes
GROUP BY diabetes;
```
