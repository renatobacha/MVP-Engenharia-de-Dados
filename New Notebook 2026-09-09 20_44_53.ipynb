```python
# Script de Profiling de Qualidade e Integridade de Dados
from pyspark.sql import functions as F

df_raw = spark.table("mvp.staging.bronze_diabetes_raw")
df_silver = spark.table("mvp.staging.silver_diabetes_clean")

# 1. Validação de Volumetria
qtd_raw = df_raw.count()
qtd_silver = df_silver.count()
print(f"[AUDITORIA] Registros Ingeridos (Bronze): {qtd_raw}")
print(f"[AUDITORIA] Registros Sanados (Silver): {qtd_silver}")
print(f"[AUDITORIA] Registros Eliminados (Ruído/Duplicatas): {qtd_raw - qtd_silver}")

# 2. Mapeamento de Nulos por Coluna
print("[AUDITORIA] Contagem de Valores Nulos na Camada Silver:")
df_silver.select([F.sum(F.col(c).isNull().cast("int")).alias(c) for c in df_silver.columns]).show()

# 3. Validação de Unicidade e Consistência Categórica
df_silver.groupBy("family_history_diabetes", "diabetes").count().show()
```
